```python
import os
import urllib.request
import pandas as pd
import numpy as np

print("Library berhasil diimpor!")
```

## 1. Konfigurasi Sumber Data
Dataset utama diambil dari agregasi harian kasus COVID-19 tingkat provinsi di Indonesia yang bersumber dari Satgas COVID-19 RI / Kementerian Kesehatan RI.

```python
DATA_URL = "https://raw.githubusercontent.com/rhmdziz/visualiasi-data-covid-19-indonesia/main/covid_19_indonesia_time_series_all.csv"
RAW_DATA_PATH = os.path.join("..", "data", "raw_data.csv")
os.makedirs(os.path.dirname(RAW_DATA_PATH), exist_ok=True)

print(f"Target URL: {DATA_URL}")
print(f"Target Simpan: {RAW_DATA_PATH}")
```

## 2. Pengunduhan Data Mentah (Fetching Data)

```python
if not os.path.exists(RAW_DATA_PATH):
    print("Mengunduh file raw_data.csv...")
    urllib.request.urlretrieve(DATA_URL, RAW_DATA_PATH)
    print("Unduhan selesai!")
else:
    print("File raw_data.csv sudah tersedia secara lokal.")
```

## 3. Inspeksi Struktur Data Mentah

```python
df_raw = pd.read_csv(RAW_DATA_PATH)
print(f"Dimensi Dataset Mentah: {df_raw.shape[0]} baris x {df_raw.shape[1]} kolom")
df_raw.head()
```

```python
print("Daftar Kolom Dataset Mentah:")
for i, col in enumerate(df_raw.columns, 1):
    print(f"{i:2d}. {col} ({df_raw[col].dtype})")
```

## 4. Evaluasi Awal Kondisi Data
Memeriksa rentang tanggal, jumlah provinsi unik, dan persentase missing values.

```python
print("Rentang Tanggal Mentah:", df_raw['Date'].min(), "s/d", df_raw['Date'].max())
print("Tingkat Wilayah:", df_raw['Location Level'].value_counts())
print("\nTotal Missing Values per Kolom (Top 10):")
print(df_raw.isnull().sum().sort_values(ascending=False).head(10))
print(f"\nTotal Duplikasi Baris Mentah: {df_raw.duplicated().sum()}")
```
