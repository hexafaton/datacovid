## 1. Persiapan dan Impor Pustaka (Library)

- `pandas` dipakai untuk mengelola dan menyusun struktur data berbentuk tabel (*DataFrame*).
- `numpy` dimanfaatkan untuk menghasilkan angka acak sebagai sampel data simulasi.
- `matplotlib.pyplot` berfungsi sebagai pustaka utama dalam menggambar dan mengatur visualisasi plot.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

## 2. Pembangkitan Data Simulasi Usia dan Penghasilan

Pengaturan nilai `seed(42)` diterapkan supaya urutan angka acak tetap sama setiap kali sel dieksekusi. Variabel usia disimulasikan berada pada kisaran 20 hingga 59 tahun, sementara nilai penghasilan dihitung melalui perkalian usia dengan faktor 500 yang ditambahkan simpangan acak (*random noise*).

```python
# Membangkitkan sampel data simulasi untuk variabel usia dan penghasilan
np.random.seed(42)  # Mengunci nilai seed agar output angka acak selalu tetap

usia = np.random.randint(20, 60, 100)
penghasilan = (usia * 500) + np.random.randn(100) * 3000  # Pola relasi linear disertai komponen acak

# Memasukkan hasil simulasi ke dalam struktur DataFrame Pandas
df_usia_penghasilan = pd.DataFrame({
    'Usia': usia,
    'Penghasilan': penghasilan
})

# Mencetak 5 sampel baris teratas dari tabel
df_usia_penghasilan.head()
```

## 3. Visualisasi Data dengan Scatter Plot

Pada bidang koordinat, sumbu mendatar (X) merepresentasikan **usia**, sementara sumbu tegak (Y) menggambarkan besaran **penghasilan**. Tiap noktah (titik) pada diagram melambangkan satu sampel individu.

```python
# Menampilkan grafik sebar (scatter plot) dari DataFrame
plt.figure(figsize=(10, 6))
plt.scatter(
    df_usia_penghasilan['Usia'],
    df_usia_penghasilan['Penghasilan'],
    color='purple',
    alpha=0.6,
    edgecolors='white'
)

plt.title('Hubungan Antara Usia dan Penghasilan')
plt.xlabel('Usia')
plt.ylabel('Penghasilan')
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.tight_layout()
plt.show()
```

## 4. Ringkasan Hasil Pengamatan

Diagram *scatter plot* sangat efektif untuk mengamati kecenderungan korelasi di antara dua variabel bertipe numerik. Berdasarkan grafik di atas, sebaran titik memperlihatkan tren positif di mana **tingkat penghasilan cenderung bertambah mengikuti kenaikan usia**, meskipun titik-titiknya menyebar di sekitar jalur linear akibat adanya faktor variasi acak (*noise*).
